```python
# Sample text
text = "I love programming. I love Python. Programming is fun. I love c++. I hate programming"

# Tokenize and preprocess
tokens = word_tokenize(text.lower())

# Generate bigrams
bigrams = list(ngrams(tokens, 2))

# Count frequencies
bigram_freq = Counter(bigrams)
unigram_freq = Counter(tokens)

# Predict the next word
def predict_next_word(word, bigram_freq, unigram_freq):
    candidates = {pair[1]: freq for pair, freq in bigram_freq.items() if pair[0] == word}
    if not candidates:
        return None
    return max(candidates)

# Test the prediction
context_word = "Hate"
next_word = predict_next_word(context_word.lower(), bigram_freq, unigram_freq)
print(f"The predicted next word for '{context_word}' is: {next_word}")
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag
from nltk.chunk import RegexpParser

# Download required NLTK data
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')

# Sample corpus
text = "The quick brown fox jumps over the lazy dog."

# Tokenize the text
tokens = word_tokenize(text)

# POS tagging
pos_tags = pos_tag(tokens)
print("POS Tags:")
print(pos_tags)

# Define a pattern for extracting noun phrases (NP)
grammar = "NP: {<DT>?<JJ>*<NN>}"

# Create a parser using the grammar
chunk_parser = RegexpParser(grammar)

# Parse the POS-tagged tokens
tree = chunk_parser.parse(pos_tags)

# Display the parse tree
print("\nParse Tree:")
print(tree)


```

```python
import numpy as np
import pandas as pd
import gensim
from gensim import corpora
from gensim.models import ldamodel
from gensim.parsing.preprocessing import preprocess_string
doc=["Machine learning is a subset of artificial intelligence.",
"Data privacy and security are important in modern technology.",
"Mobile apps are becoming more popular among users.",
"E-commerce platforms make online shopping convenient.",
"Cyber security measures are crucial to protect against cyber threats."]
process_doc=[]
for d in doc:
  process_doc.append(preprocess_string(d))
dic=corpora.Dictionary(process_doc)
courpus=[dic.doc2bow(doc) for doc in process_doc]
model=ldamodel.LdaModel(courpus,num_topics=2,id2word=dic,passes=15)
topics=model.print_topics(5)
for topic in topics:
  print(topic)
```

```python
import gensim
import numpy as np
from gensim.models import Word2Vec
corpus = [
"Machine learning is a subset of artificial intelligence.",
"Data privacy and security are important in modern technology.",
"Mobile apps are becoming more popular among users.",
"E-commerce platforms make online shopping convenient.",
"Cyber security measures are crucial to protect against cyber threats."
]
data = [sentence.lower().split() for sentence in corpus]
model = Word2Vec(data, min_count=1)
model
for sentence in corpus:
  vector = np.mean([model.wv[word] for word in sentence.lower().split() if word in model.wv],axis=1)
  print(f"Sentence: {sentence}\nAssignment Vector (first 5 values): {vector[:5]}\n")
```

```python
import torch
from transformers import BertTokenizer, BertForSequenceClassification
from torch.utils.data import DataLoader
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Sample data
texts = ["I love programming!", "This is a bad experience.", "The movie was fantastic!", "I dislike the weather today."]
labels = [1, 0, 1, 0]

# Tokenizer and model
tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
model = BertForSequenceClassification.from_pretrained("bert-base-uncased", num_labels=2).to("cuda" if torch.cuda.is_available() else "cpu")

# Tokenization function
def tokenize_data(texts, labels):
    encodings = tokenizer(texts, padding=True, truncation=True, max_length=128, return_tensors="pt")
    dataset = torch.utils.data.TensorDataset(encodings["input_ids"], encodings["attention_mask"], torch.tensor(labels))
    return DataLoader(dataset, batch_size=2, shuffle=True)

# Prepare data loaders
train_texts, val_texts, train_labels, val_labels = train_test_split(texts, labels, test_size=0.2, random_state=42)
train_loader = tokenize_data(train_texts, train_labels)
val_loader = tokenize_data(val_texts, val_labels)

# Training setup
optimizer = torch.optim.Adam(model.parameters(), lr=5e-5)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Training loop
for epoch in range(2):
    model.train()
    for input_ids, attention_mask, labels in train_loader:
        optimizer.zero_grad()
        outputs = model(input_ids.to(device), attention_mask.to(device), labels=labels.to(device))
        outputs.loss.backward()
        optimizer.step()

# Evaluation
model.eval()
val_preds, val_labels = [], []
with torch.no_grad():
    for input_ids, attention_mask, labels in val_loader:
        logits = model(input_ids.to(device), attention_mask.to(device)).logits
        val_preds.extend(torch.argmax(logits, dim=1).cpu().numpy())
        val_labels.extend(labels.numpy())

# Accuracy
print(f"Validation Accuracy: {accuracy_score(val_labels, val_preds):.4f}")
```
